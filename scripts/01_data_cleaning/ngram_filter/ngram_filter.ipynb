```python
import pandas as pd
import os

file_path = 'ngram_demo_input.xlsx'
df = pd.read_excel(file_path)

output_folder = './'

file_path_noise = 'data/noise_terms.xlsx'
df_noise = pd.read_excel(file_path_noise)

noise_words = df_noise['三級分類'].dropna().unique().tolist()

def check_noise(text):
    if pd.isna(text):
        return "clean"
    
    text_str = str(text)

    for word in noise_words:
        if str(word) in text_str:
            return "fuzzy" # Data is deemed fuzzy if any noise word is matched.
    return "clean"


df['clean_flag'] = df['摘要'].apply(check_noise)

fuzzy_count = len(df[df['clean_flag'] == 'fuzzy'])
total_count = len(df)
interference_rate = (fuzzy_count / total_count) * 100

print(f"total: {total_count}")
print(f"fuzzy: {fuzzy_count}")
print(f"noise rate: {interference_rate:.2f}%")

output_path = os.path.join(output_folder, 'ngram_demo_output.xlsx')
df.to_excel(output_path, index=False)
```
